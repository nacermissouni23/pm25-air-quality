```python
import pandas as pd

df = pd.read_csv('GlobalWeatherRepository.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
df['date'] = pd.to_datetime(df['last_updated']).dt.date.drop(columns=['last_updated'])
df['date'].head()
```

```python
unique_dates = pd.DataFrame({'date': df['date'].unique()})
unique_dates.to_csv('unique_dates.csv', index=False)
unique_dates.head()
```

```python
unique_df = df.groupby(['latitude', 'longitude']).first().reset_index()[['latitude', 'longitude']]
unique_df.to_csv('unique_locations.csv', index=False)
print (f'shape: {unique_df.shape}')
unique_df.head()
```

```python
mask = (df['latitude'] == -41.3000) & (df['longitude'] == 174.7800)
dates = pd.to_datetime(df.loc[mask, 'date'])  # Convert to datetime

print(f"Start: {dates.min().date()}")
print(f"End: {dates.max().date()}")
```
